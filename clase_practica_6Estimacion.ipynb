¿Qué es un parámetro?

Un parámetro es una medida numérica que describe una característica de toda la población, como la media (μ), la varianza (σ
2
) o la desviación estándar (σ). Generalmente es desconocido.

¿Qué es un estadístico?

Un estadístico es una medida calculada a partir de una muestra y utilizada para estimar un parámetro poblacional.

¿Qué es una estimación puntual?

Una estimación puntual consiste en utilizar un único valor obtenido de una muestra para aproximar el valor de un parámetro poblacional.

#Estimación puntual

```r
###############################################################
# ESTIMACIÓN PUNTUAL
###############################################################

# Objetivo:
# Estimar la media, la varianza y el desvío estándar poblacional
# a partir de una muestra de datos.

###############################################################
# Cargar el conjunto de datos
###############################################################

data("airquality")

# Visualizar las primeras observaciones
head(airquality)

###############################################################
# Seleccionar la variable de estudio
###############################################################

# Velocidad del viento (mph)
viento <- airquality$Wind

###############################################################
# Tamaño de la muestra
###############################################################

n <- length(viento)

cat("Tamaño de la muestra:", n, "\n\n")

###############################################################
# Estimación puntual de la media
###############################################################

media <- mean(viento)

cat("Estimación puntual de la media:", round(media,2), "mph\n")

###############################################################
# Estimación puntual de la varianza
###############################################################

varianza <- var(viento)

cat("Estimación puntual de la varianza:", round(varianza,2), "\n")

###############################################################
# Estimación puntual del desvío estándar
###############################################################

desvio <- sd(viento)

cat("Estimación puntual del desvío estándar:", round(desvio,2), "mph\n")

###############################################################
# Resumen descriptivo
###############################################################

summary(viento)

###############################################################
# Histograma
###############################################################

hist(viento,
     col="lightblue",
     border="white",
     probability=TRUE,
     main="Estimación puntual de la velocidad del viento",
     xlab="Velocidad del viento (mph)",
     ylab="Densidad")

# Línea de la media
abline(v=media,
       col="red",
       lwd=3)

# Etiqueta de la media
text(media,
     par("usr")[4]*0.90,
     labels=paste("Media =",round(media,2)),
     pos=4,
     col="red")

###############################################################
# Diagrama de caja
###############################################################

boxplot(viento,
        horizontal=TRUE,
        col="lightgreen",
        main="Diagrama de caja de la velocidad del viento",
        xlab="Velocidad del viento (mph)")

abline(v=media,
       col="red",
       lwd=2)

###############################################################
# Tabla resumen
###############################################################

tabla_resultados <- data.frame(
  Estadistico=c("Tamaño de muestra",
                "Media",
                "Varianza",
                "Desvío estándar"),
  Valor=c(n,
          round(media,2),
          round(varianza,2),
          round(desvio,2))
)

tabla_resultados

###############################################################
# Interpretación
###############################################################

cat("\nINTERPRETACIÓN\n")
cat("--------------------------------------------------\n")

cat("La media muestral es una estimación puntual de la media poblacional.\n")
cat("En este caso, se estima que la velocidad promedio del viento es de",
    round(media,2),"mph.\n\n")

cat("La varianza muestral estima la variabilidad de la población.\n")
cat("Un valor de", round(varianza,2),
    "indica que existe dispersión entre las velocidades observadas.\n\n")

cat("El desvío estándar es de", round(desvio,2),
    "mph, lo que significa que, en promedio, las observaciones se alejan aproximadamente esa cantidad respecto de la media.\n\n")

cat("Estas medidas constituyen estimaciones puntuales porque se obtiene un único valor para estimar cada parámetro poblacional.\n")
```

# **Estimacion por intervalos de confianza**

INTERVALO DE CONFIANZA PARA LA MEDIA (σ CONOCIDA)

```r
###############################################################
# INTERVALO DE CONFIANZA PARA LA MEDIA (σ CONOCIDA)-Metodo extendido
##############################################################
###############################################################
# Datos de la muestra
###############################################################

media <- 2.6          # Media muestral
desviacion <- 0.3     # Desviación estándar poblacional conocida (σ)
n <- 36               # Tamaño de muestra

###############################################################
# Nivel de confianza
###############################################################

confianza <- 0.95

alpha <- 1-confianza

###############################################################
# Valor crítico Z
###############################################################

z <- qnorm(1-alpha/2)

cat("Valor crítico Z =", round(z,3), "\n")

###############################################################
# Error estándar
###############################################################

error_estandar <- desviacion/sqrt(n)

cat("Error estándar =", round(error_estandar,4), "\n")

###############################################################
# Intervalo de confianza
###############################################################

limite_inferior <- media-z*error_estandar

limite_superior <- media+z*error_estandar


cat("\nIntervalo de confianza del 95%:\n")
cat("[",
    round(limite_inferior,3),
    ",",
    round(limite_superior,3),
    "]\n")


###############################################################
# Método abreviado con paquete BSDA
###############################################################
install.packages("BSDA")
library(BSDA)

zsum.test(mean.x=media,
          n.x=n,
          sigma.x=desviacion,
          conf.level=0.95)


###############################################################
# Interpretación
###############################################################

cat("\nINTERPRETACIÓN\n")
cat("--------------------------------------\n")

cat("Con un nivel de confianza del 95%, se estima que la media poblacional\n")
cat("se encuentra entre",
    round(limite_inferior,3),
    "y",
    round(limite_superior,3),
    ".\n\n")

cat("Esto significa que si se tomaran muchas muestras similares y se calcularan\n")
cat("intervalos de confianza del mismo modo, aproximadamente el 95% de ellos\n")
cat("contendrían el verdadero valor de la media poblacional.\n")
```

Intervalo de confianza para μ cuando se desconoce σ y n<30:
#(Una sola muestra: estimación de la media)

```r
###############################################################
# INTERVALO DE CONFIANZA PARA LA MEDIA
# σ DESCONOCIDA Y n < 30
# Distribución t de Student
###############################################################
###############################################################
# Datos de la muestra
###############################################################

datos <- c(9.8, 10.2, 10.4, 9.8,
           10.0, 10.2, 9.6)

###############################################################
# Estadísticos muestrales
###############################################################

media_datos <- mean(datos)

s_datos <- sd(datos)

n_datos <- length(datos)

gl <- n_datos - 1


cat("Media muestral:",
    round(media_datos,3),"\n")

cat("Desvío estándar muestral:",
    round(s_datos,3),"\n")

cat("Tamaño de muestra:",
    n_datos,"\n")

cat("Grados de libertad:",
    gl,"\n")


###############################################################
# Nivel de confianza
###############################################################

confianza <- 0.95

alpha <- 1-confianza


###############################################################
# Valor crítico t
###############################################################

t_critico <- qt(1-alpha/2,
                df=gl)

cat("\nt crítico:",
    round(t_critico,3),"\n")


###############################################################
# Error estándar
###############################################################

error_estandar <- s_datos/sqrt(n_datos)

cat("Error estándar:",
    round(error_estandar,4),"\n")


###############################################################
# Intervalo de confianza
###############################################################

limite_inferior <- media_datos -
                   t_critico*error_estandar

limite_superior <- media_datos +
                   t_critico*error_estandar


cat("\nIntervalo de confianza del 95%:\n")

cat("[",
    round(limite_inferior,3),
    ",",
    round(limite_superior,3),
    "]\n")


###############################################################
# Método abreviado con paquete BSDA
###############################################################

library(BSDA)

tsum.test(mean.x = media_datos,
          s.x = s_datos,
          n.x = n_datos,
          conf.level = 0.95)



###############################################################
# Interpretación
###############################################################

cat("\nINTERPRETACIÓN\n")
cat("--------------------------------------\n")

cat("Con un nivel de confianza del 95%, se estima que\n")
cat("la verdadera media poblacional se encuentra entre",
    round(limite_inferior,3),
    "y",
    round(limite_superior,3),
    ".\n\n")

cat("Se utiliza la distribución t de Student porque:\n")
cat("- La muestra es pequeña (n < 30).\n")
cat("- La desviación estándar poblacional es desconocida.\n")
```

# Intervalos de confianza de la variancia poblacional.

```r
###############################################################
# INTERVALO DE CONFIANZA PARA LA VARIANZA POBLACIONAL
# Distribución Chi-cuadrado
##############################################################
###############################################################
# Datos de la muestra
###############################################################

litros <- c(9.8, 10.2, 10.4, 9.8,
            10.0, 10.2, 9.6)


###############################################################
# Estadísticos muestrales
###############################################################

n <- length(litros)

s2 <- var(litros)

gl <- n-1


cat("Tamaño de muestra:", n,"\n")

cat("Varianza muestral:",
    round(s2,4),"\n")

cat("Grados de libertad:",
    gl,"\n")


###############################################################
# Nivel de confianza
###############################################################

confianza <- 0.95

alpha <- 1-confianza


###############################################################
# Valores críticos Chi-cuadrado
###############################################################

chi2_inferior <- qchisq(alpha/2,
                        df=gl)

chi2_superior <- qchisq(1-alpha/2,
                        df=gl)


cat("\nChi-cuadrado inferior:",
    round(chi2_inferior,3),"\n")

cat("Chi-cuadrado superior:",
    round(chi2_superior,3),"\n")


###############################################################
# Intervalo de confianza para la varianza
###############################################################

limite_inferior <- ((gl)*s2)/
                   chi2_superior


limite_superior <- ((gl)*s2)/
                   chi2_inferior


cat("\nIntervalo de confianza del 95% para la varianza:\n")

cat("[",
    round(limite_inferior,4),
    ",",
    round(limite_superior,4),
    "]\n")


###############################################################
# Método abreviado
###############################################################
install.packages("EnvStats")
library(EnvStats)

varTest(litros,
        conf.level=0.95)


###############################################################
# Interpretación
###############################################################

cat("\nINTERPRETACIÓN\n")
cat("--------------------------------------\n")

cat("Con un nivel de confianza del 95%, se estima que\n")
cat("la varianza poblacional se encuentra entre",
    round(limite_inferior,4),
    "y",
    round(limite_superior,4),
    ".\n\n")

cat("Este intervalo representa la incertidumbre asociada\n")
cat("a la estimación de la variabilidad poblacional.\n")

cat("Se utiliza la distribución Chi-cuadrado porque\n")
cat("el parámetro estimado es la varianza poblacional.\n")
```

 # Intervalo de confianza para una proporción poblacional

```r
###############################################################
# INTERVALO DE CONFIANZA PARA UNA PROPORCIÓN POBLACIONAL
###############################################################
###############################################################
# Datos
###############################################################

x <- 55       # Número de éxitos o individuos con la característica
n <- 100      # Tamaño de muestra

# Proporción muestral

p_muestral <- x/n

cat("Proporción muestral:",
    p_muestral,
    "\n")


###############################################################
# Nivel de confianza
###############################################################

confianza <- 0.95

alpha <- 1-confianza


###############################################################
# Valor crítico Z
###############################################################

z <- qnorm(1-alpha/2)

cat("Valor crítico Z:",
    round(z,3),
    "\n")


###############################################################
# Error estándar
###############################################################

error_estandar <- sqrt((p_muestral*(1-p_muestral))/n)


cat("Error estándar:",
    round(error_estandar,4),
    "\n")


###############################################################
# Intervalo de confianza
###############################################################

limite_inferior <- p_muestral -
                   z*error_estandar


limite_superior <- p_muestral +
                   z*error_estandar


cat("\nIntervalo de confianza del 95%:\n")

cat("[",
    round(limite_inferior,3),
    ",",
    round(limite_superior,3),
    "]\n")


###############################################################
# Método abreviado con R
###############################################################

prop.test(x=55,
          n=100,
          conf.level=0.95)


###############################################################
# Interpretación
###############################################################

cat("\nINTERPRETACIÓN\n")
cat("--------------------------------------\n")

cat("La proporción observada en la muestra es",
    round(p_muestral*100,2),
    "%.\n\n")

cat("Con un nivel de confianza del 95%, se estima que\n")
cat("la proporción poblacional verdadera se encuentra entre",
    round(limite_inferior*100,2),
    "% y",
    round(limite_superior*100,2),
    "%.\n\n")

cat("Esto significa que, si se repitiera el procedimiento\n")
cat("con muchas muestras similares, aproximadamente el 95%\n")
cat("de los intervalos obtenidos contendrían la verdadera proporción poblacional.\n")
```

Análisis de normalidad
Prueba de Shapiro-Wilk

Hipótesis:

H0: Los datos siguen una distribución normal

H1:Los datos no siguen una distribución normal

Regla de decision:
Si:
p>0.05 No se rechaza H0, por lo tanto los datos pueden considerarse normales.
Si:
p<0.05 Se rechaza H0, indicando ausencia de normalidad.

```r
###############################################################
# Normalidad analítica
###############################################################

shapiro.test(datosCompleto$Definitiva)

shapiro.test(datosCompleto$Gastos)

shapiro.test(datosCompleto$Ingreso)
```
