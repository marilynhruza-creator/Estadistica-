# **Regresión lineal simple**

¿Que podemos interpretar de esto?
Si el coeficiente asociado a costos es positivo, significa que existe una relación directa: al aumentar los costos, las ventas tienden a incrementarse.
La regresion nos permite "predecir" como va a actuar una variable con respecto a la otra.

```r
######################################################
## REGRESIÓN LINEAL SIMPLE
######################################################

# Datos

costos <- c(40,20,25,20,30,50,40,20,50,40,25,50)

ventas <- c(385,400,395,365,475,440,490,
            420,560,525,480,510)

datos <- data.frame(costos,ventas)
```

```r
######################################################
# Exploración gráfica
######################################################

pairs(datos)
```

El código pairs(datos) genera una matriz de gráficos de dispersión para todas las variables. Permite visualizar rápidamente las relaciones bivariadas entre cada par de variables.

```r
######################################################
# Matriz de correlación
######################################################

cor(datos)
```

El código cor(datos) calcula la matriz de correlación para el data frame llamado datos. Esta función se utiliza para determinar la relación lineal entre todos los pares de variables.
*En este caso el resultado 0.63 nos esta diciendo que existe una relacion que es positiva (directamente proporcional) y moderadamente alta.*

```r
######################################################
# Modelo de regresión
######################################################

modelo_simple <- lm(ventas ~ costos,
                    data = datos)
```

La función lm() (linear model) se utiliza para ajustar modelos lineales. *En este caso queremos modelar 'ventas' como la variable dependiente y 'costos' como la variable independiente (explicativa).*

```r
######################################################
# Resultados del modelo
######################################################

summary(modelo_simple)


#funcion de Regresion obtenida Y (ventas) = 343,706 +3,221*costos
#Coeficiente de determinacion 0.403
```

El código summary(modelo_simple) muestra un resumen detallado del modelo de regresión lineal simple que hemos creado, modelo_simple.

Puedes usar este modelo_simple para hacer predicciones. Por ejemplo, si tuvieras nuevos datos de 'costos', podrías predecir las 'ventas' correspondientes reemplazando el valor del costo en la funcion de regresion obtenida. En R podemos utilizar la funcion predict().

```r
######################################################
# Gráfico de dispersión y recta de regresión
######################################################

plot(datos$costos,
     datos$ventas,
     pch=19,
     col="blue",
     xlab="Costos",
     ylab="Ventas",
     main="Regresión lineal simple")


abline(modelo_simple,
       col="red",
       lwd=2)
```

# **Regresion lineal multiple**

```r
######################################################
## REGRESIÓN LINEAL MÚLTIPLE
######################################################

# Datos

p_final <- c(95,77,80,100,97,
             70,50,80,92,84)

p_inicial <- c(42,33,33,45,39,
               36,32,41,40,38)

p_alimento <- c(272,226,259,292,311,
                183,173,236,230,235)


crecimiento <- data.frame(
  p_final,
  p_inicial,
  p_alimento
)


######################################################
# Exploración
######################################################

pairs(crecimiento)


######################################################
# Correlaciones
######################################################

cor(crecimiento)


######################################################
# Modelo múltiple
######################################################

modelo_multiple <- lm(
  p_final ~ p_inicial + p_alimento,
  data = crecimiento
)


######################################################
# Resultados
######################################################

summary(modelo_multiple)

#Y(Peso_final)= -22,99 + 1,39*p_inicial+0,21*p_alimento
#R2  = 0,87
```

```r
# Cuatro gráficos básicos de diagnóstico para regresión múltiple

par(mfrow = c(2, 2))
plot(modelo_multiple)
```

Estos cuatro gráficos permiten evaluar:

1. Residuales vs ajustados

Permite verificar si existe una relación lineal adecuada.

2. QQ Plot

Evalúa normalidad de los residuos.

3. Scale Location

Evalúa homogeneidad de varianzas.

4. Residuales vs leverage

Detecta observaciones influyentes.

**La regresión lineal permite modelar y cuantificar la relación entre una variable respuesta y una o más variables explicativas. La correlación mide la intensidad de la relación lineal, mientras que la regresión permite construir un modelo matemático para explicar o predecir el comportamiento de la variable respuesta.**
