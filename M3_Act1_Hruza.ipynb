```python
def clasificar_precipitacion(pp_mm):
    """Clasifica el clima según la precipitación anual en mm."""
    if pp_mm < 200:
        return "árido"
    elif pp_mm <= 500:
        return "semiárido"
    elif pp_mm <= 700:
        return "subhúmedo seco"
    elif pp_mm <= 1000:
        return "subhúmedo húmedo"
    else:
        return "húmedo"
```

## Ejercicio 1  Condicionales

```python
print("San Juan (100 mm):", clasificar_precipitacion(100))
print("Mendoza (220 mm):", clasificar_precipitacion(220))
print("Bahía Blanca (600 mm):", clasificar_precipitacion(600))
print("Córdoba (750 mm):", clasificar_precipitacion(750))
print("Buenos Aires (1200 mm):", clasificar_precipitacion(1200))
print("Posadas (1700 mm):", clasificar_precipitacion(1700))
```

## Ejercicio 2 – Bucle while

```python
superficie = 100      # km² deforestados al inicio
anio = 0              # contador de años
limite = 500          # km² que queremos superar

print(f"Año {anio}: {superficie:.2f} km²")

while superficie <= limite:
    superficie = superficie * 1.08   # crece un 8%
    anio += 1                        # pasó un año más
    print(f"Año {anio}: {superficie:.2f} km²")

print(f"\nSe superan los {limite} km² después de {anio} años.")
```

## Ejercicio 3 – Comprensión de listas.

```python
# Lista de áreas en km²
areas = [1250, 80, 560, 3200, 450, 900, 1800, 75]

# 1) Áreas convertidas a hectáreas (1 km² = 100 ha)
areas_ha = [a * 100 for a in areas]

# 2) Solo las áreas mayores a 1000 km²
mayores_1000 = [a for a in areas if a > 1000]

# 3) Strings con formato según si supera o no los 500 km²
etiquetas = [f"{a} km² (grande)" if a > 500 else f"{a} km² (pequeña)" for a in areas]

print("Áreas en ha:", areas_ha)
print("Mayores a 1000 km²:", mayores_1000)
print("Etiquetas:")
for e in etiquetas:
    print("  ", e)
```
