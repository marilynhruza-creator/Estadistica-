Operador Igualdad

```python
var = 0  # Asignamos 0 a var
print(var == 0)

var = 1  # Asignamos 1 a var
print(var == 0)
```

Operador Desigualdad

```python
var = 0  # Asignamos 0 a var
print(var != 0)

var = 1  # Asignamos 1 a var
print(var != 0)
```

Condicional If

```python
edad = int(input("Ingrese edad"))
if edad <= 18:
  print('Menor')
elif edad > 65:
  print('Jubilado')
else:
  print('Activo')

```

Bucle While

```python
# Pregunta al usuario por un número hasta que introduce 0.
num = None
while num != 0:
  num = int(input('Introduce un número: '))
```

```python
# Pregunta al usuario por un número hasta que introduce 0.
while True:
  num = int(input('Introduce un número: '))
  if num == 0:
    break
```

Bucle Iterativo

```python
palabra = (input("ingrese su apellido: "))
for l in palabra:
  print(l)
```

```python
año = int(input("Ingrese un año: "))

if año < 1582:
	print("No está dentro período del calendario gregoriano")
else:
  if año % 4 != 0:
    print("Año común")
  elif año % 100 != 0:
    print("Año bisiesto")
  elif año % 400 != 0:
    print("Año común")
  else:
    print("Año bisiesto")
```

```python
numero = int(input("Ingrese un número: "))
if numero < 0:
  print ("Negativo")
elif numero > 0:
  print ("Positivo")
else:
  print ("Cero")
```
