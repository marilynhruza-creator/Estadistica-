La función sample() en R para crear una muestra de la velocidad del viento:

*   `replace = TRUE`: Especifica que el muestreo se realiza **con reemplazo**. Esto significa que un mismo valor de velocidad del viento puede ser seleccionado múltiples veces en la muestra.

Mínimo (Min.): El valor más pequeño en la muestra.
1er Cuartil (1st Qu.): El valor por debajo del cual se encuentra el 25% de los datos.
Mediana (Median): El valor central de la muestra (el 50% de los datos está por debajo y el 50% por encima).
Media (Mean): El promedio de todos los valores en la muestra.
3er Cuartil (3rd Qu.): El valor por debajo del cual se encuentra el 75% de los datos.
Máximo (Max.): El valor más grande en la muestra.

# **Visualización de datos**

```r
install.packages("dplyr")
```

La función install.packages() en R para instalar el paquete dplyr.

```r
library("dplyr")
```

dplyr es una librería muy popular que proporciona un conjunto de herramientas para manipular datos de forma eficiente y legible.
La función library () permite el uso de la libreria recientemente instalada y esta debe usarse antes de utilizar las herramientas de dplyr o cualquier otra libreria previamente instalada que no forme parte de la libreria base de R.

```r
tabla <- calidad_aire|>
  group_by(Month) |> #operador pipe |>
  summarise(
    n = n(),
    suma= sum(Wind),
    maximo=max(Wind),
    minimo=min(Wind))
```

Este código R utiliza el paquete dplyr para realizar un análisis de resumen de los datos de calidad del aire. Aquí te explico qué hace cada parte:

tabla <- calidad_aire |>: Comienza con el conjunto de datos calidad_aire y usa el operador pipe (|>) para pasar este conjunto de datos a la siguiente función. El resultado final de esta serie de operaciones se guardará en una nueva variable llamada tabla.
group_by(Month): Esta función agrupa los datos por la columna Month (Mes). Esto significa que las operaciones posteriores se realizarán de forma separada para cada mes.
summarise(...): Después de agrupar, la función summarise() calcula estadísticas resumidas para cada grupo (es decir, para cada mes).
n = n(): Calcula el número de observaciones (n) en cada grupo (cuántos registros hay para cada mes).
suma = sum(Wind): Calcula la suma total de la velocidad del viento (Wind) para cada mes.
maximo = max(Wind): Encuentra el valor máximo de la velocidad del viento (Wind) registrado en cada mes.
minimo = min(Wind): Encuentra el valor mínimo de la velocidad del viento (Wind) registrado en cada mes.
En resumen, este código crea una tabla (tabla) que muestra el recuento de días, la suma total del viento, el viento máximo y el viento mínimo, todo esto calculado para cada mes presente en el conjunto de datos calidad_aire

ggplot2 es una librería muy popular en R que proporciona un sistema para crear visualizaciones de datos declarativas, elegantes y muy personalizables.

```r
library("ggplot2")
```

library("ggplot2"), se utiliza para cargar el paquete ggplot2. Previo a uso se debe realizar su intalacion utilizando el codigo install.packages ("ggplot2")

```r
ggplot(data = iris) + aes(x = Species, y = Sepal.Width) +
  geom_point() +
  labs(x = "Species", y = "Sepal width (cm)", title = "Gráfico de dispersión")
```

Este código R utiliza la librería ggplot2 para crear un gráfico de dispersión a partir del conjunto de datos iris. Te explico cada parte:

ggplot(data = iris): Inicializa el gráfico, especificando que los datos a utilizar son del conjunto iris.
+ aes(x = Species, y = Sepal.Width): Define las estéticas del gráfico. Aquí se indica que la variable Species (Especies) se mapeará al eje X, y Sepal.Width (Ancho del Sépalo) al eje Y.
+ geom_point(): Añade una capa de puntos al gráfico, lo que resulta en un gráfico de dispersión.
+ labs(x = "Species", y = "Sepal width (cm)", title = "Gráfico de dispersión"): Establece las etiquetas y el título del gráfico. Se nombra el eje X como "Species", el eje Y como "Sepal width (cm)" y el título del gráfico como "Gráfico de dispersión".

```r
ggplot(data = iris) + aes(x = Species, y = Petal.Length) +
  geom_boxplot(coef = 3) +
  labs(x = "Species", y = "Petal length (cm)", title = "Boxplot")
```

A diferencia del codigo anterior este utiliza la funcion:
+ geom_boxplot(coef = 3): que añade una capa de diagramas de cajas al gráfico. El argumento coef = 3 es un parámetro que controla la longitud de los bigotes del diagrama de cajas; un valor de 3 extiende los bigotes hasta 3 veces el rango intercuartílico (IQR) más allá de los cuartiles, lo que puede ayudar a identificar valores atípicos (outliers) de manera más laxa.
+ labs(x = "Species", y = "Petal length (cm)", title = "Boxplot"): Establece las etiquetas y el título del gráfico. Se nombra el eje X como "Species", el eje Y como "Petal length (cm)" y el título del gráfico como "Boxplot".

```r
ggplot(data = iris) + aes(x = Petal.Length, fill = Species) +
  geom_histogram() +
  labs(x = "Petal length (cm)", fill = "Species", title = "Histograma")
```

A diferencia del primer codigo de ggplot visto, este utiliza la funcion:
+ geom_histogram(): Añade una capa de histograma al gráfico, lo que muestra la distribución de Petal.Length para cada especie.
+ labs(x = "Petal length (cm)", fill = "Species", title = "Histograma"): Establece las etiquetas y el título del gráfico. Se nombra el eje X como "Petal length (cm)", la leyenda de relleno como "Species" y el título del gráfico como "Histograma".

```r
ggplot(data = iris) + aes(x = "", fill = Species) +
  geom_bar(width = 1, color = "white") + coord_polar("y") +
  labs(title = "Frecuencia especies") + theme_void()
```

A diferencia del primer codigo de ggplot visto, este utiliza la funcion:
+ geom_bar(width = 1, color = "white"): Añade una capa de barras al gráfico. width = 1 asegura que la barra ocupe todo el espacio disponible. color = "white" añade un borde blanco a cada segmento, lo que ayuda a diferenciarlos.
+ coord_polar("y"): Esta es la clave para transformar el gráfico de barras en un gráfico circular (similar a un pastel o rosquilla). Proyecta las barras sobre un sistema de coordenadas polares, usando el eje y (la altura de la barra, que representa la cuenta de cada especie) para determinar el ángulo del segmento.
+ labs(title = "Frecuencia especies"): Establece el título del gráfico como "Frecuencia especies".
+ theme_void(): Aplica un tema vacío, eliminando ejes, etiquetas y elementos de fondo para que el gráfico circular sea el centro de atención.

```r
ggplot(data = iris) + aes(x = Sepal.Width, y = Sepal.Length) +
  geom_point() +
  geom_line(stat = 'smooth', method = 'lm') +
  facet_wrap(~ Species, labeller = label_both)
```

A diferencia del primer codigo de ggplot visto, este utiliza la funcion:
+ geom_point(): Añade una capa de puntos al gráfico, lo que resulta en un gráfico de dispersión mostrando los datos individuales.
+ geom_line(stat = 'smooth', method = 'lm'): Añade una línea suavizada al gráfico. stat = 'smooth' le dice a ggplot que estime una tendencia, y method = 'lm' especifica que se utilice un modelo lineal (regresión lineal) para calcular esta línea de tendencia. Esta línea muestra la relación general entre el ancho y el largo del sépalo.
+ facet_wrap(~ Species, labeller = label_both): Crea un panel separado para cada nivel de la variable Species (especie). Esto significa que tendrás un gráfico de dispersión y su línea de tendencia para 'setosa', otro para 'versicolor' y otro para 'virginica'. labeller = label_both asegura que la etiqueta de cada panel incluya el nombre de la variable ('Species') junto con el valor de la especie (por ejemplo, 'Species: setosa').

Este paquete es comúnmente utilizado para análisis estadísticos en agricultura.

```r
frecuencias<- with(iris, graph.freq(Sepal.Length, plot = TRUE))
```

La función graph.freq del paquete agricolae se utiliza para calcular y visualizar la distribución de frecuencias de la variable Sepal.Length del conjunto de datos iris.

frecuencias <- ...: Asigna el resultado de la función graph.freq a una nueva variable llamada frecuencias.
with(iris, ...): Esta función permite ejecutar una expresión dentro del entorno del conjunto de datos iris. Esto significa que Sepal.Length se refiere directamente a la columna 'Sepal.Length' de iris.
graph.freq(Sepal.Length, plot = TRUE): Esta es la función principal del paquete agricolae.
Sepal.Length: Es la variable para la cual se calculará y graficará la distribución de frecuencias.
plot = TRUE: Indica que la función debe generar y mostrar un gráfico de la distribución de frecuencias (típicamente un histograma o gráfico de barras de frecuencias).

```r
tabla_s<- print(table.freq(frecuencias),row.names=FALSE)
```

tabla_s <- ...: Asigna el resultado de la operación a una nueva variable llamada tabla_s.
table.freq(frecuencias): Esta función, también del paquete agricolae, toma el objeto frecuencias (que contiene la distribución de frecuencias calculada previamente con graph.freq) y lo convierte en una tabla de frecuencias estructurada. Esta tabla incluirá información como las clases, las frecuencias absolutas, las frecuencias relativas, etc.
print(..., row.names = FALSE): La función print() se utiliza para mostrar el contenido de la tabla de frecuencias en la consola. El argumento row.names = FALSE especifica que no se muestren los nombres de las filas en la salida impresa de la tabla, lo cual es útil para una presentación más limpia.


**Lower:** Es el límite inferior de cada intervalo de clase. Indica el valor mínimo que una observación puede tener para ser incluida en esa clase.
**Upper:** Es el límite superior de cada intervalo de clase. Indica el valor máximo que una observación puede tener para ser incluida en esa clase.
**Main: **Es el punto medio de cada intervalo de clase. Se calcula como el promedio de los límites inferior y superior de la clase (Lower + Upper) / 2.
Frequency: Representa la frecuencia absoluta, es decir, el número de observaciones (en este caso, longitudes de sépalo) que caen dentro de cada intervalo de clase.
**Percentage:** Es la frecuencia relativa porcentual. Indica el porcentaje de observaciones totales que se encuentran en cada intervalo de clase.
**CF (Cumulative Frequency):** Es la frecuencia acumulada. Para cada clase, es la suma de su frecuencia absoluta y las frecuencias absolutas de todas las clases anteriores. Muestra cuántas observaciones son menores o iguales al límite superior de esa clase.
**CPF (Cumulative Percentage Frequency):** Es la frecuencia acumulada porcentual. Para cada clase, es la suma de su porcentaje y los porcentajes de todas las clases anteriores. Muestra qué porcentaje de observaciones son menores o iguales al límite superior de esa clase.
