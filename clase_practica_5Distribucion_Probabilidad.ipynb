#Distribucion binomial

```r

#####################################################################################
##La probabilidad de que un paciente se recupere de una rara enfermedad sanguinea
#es de 0.4. Si se sabe que 15 personas contrajeron la enfermedad.
#¿Cual es la probabilidad de que:
#sobrevivan al menos 10    b) sobrevivan de 3 a 8, y  c) sobrevivan exactamente 5?

#Parámetros
n <- 15
p <- 0.4

# (a) Probabilidad de que sobrevivan al menos 10
p_s_10 <- sum(dbinom(10:15, size = n, prob = p))
p_s_10
#La probabilidad obtenida representa la posibilidad de que 10 o más personas de las 15 se recuperen.
#Matemáticamente, corresponde a la suma de las probabilidades de que se recuperen exactamente 10, 11, 12, 13, 14 o 15 personas.
#En este caso, el valor obtenido es bajo, lo que indica que es poco probable que se recuperen al menos 10 personas,

# (b) Probabilidad de que sobrevivan de 3 a 8
p_3_a_8 <- sum(dbinom(3:8, size = n, prob = p))
p_3_a_8
#Este resultado representa la probabilidad de que entre 3 y 8 personas, inclusive, se recuperen.
#Como este intervalo incluye valores cercanos al número esperado de recuperaciones ($n \times p = 15 \times 0.4 = 6$),
#la probabilidad suele ser elevada. Esto indica que lo más probable es observar una cantidad de recuperaciones cercana a 6 personas.

# (c) Probabilidad de que sobrevivan exactamente 5
p_5 <- dbinom(5, size = n, prob = p)
p_5
#Este valor corresponde a la probabilidad de que exactamente 5 personas se recuperen, ni una más ni una menos.
#Es una probabilidad puntual dentro de la distribución binomial y
#permite conocer la posibilidad de que ocurra un número específico de recuperaciones.

#grafico
# Valores posibles de éxitos
x <- 0:n

# Probabilidades para cada número de éxitos
prob <- dbinom(x, size = n, prob = p)

# Graficar
#################################
# GRAFICO
#################################

# Valores posibles de éxitos
x <- 0:n

# Probabilidades
prob <- dbinom(x, size = n, prob = p)

# Colores por defecto
colores <- rep("lightblue", length(x))

# (b) Entre 3 y 8
colores[x >= 3 & x <= 8] <- "gold"

# (a) Al menos 10
colores[x >= 10] <- "tomato"

# (c) Exactamente 5 (sobrescribe el color anterior)
colores[x == 5] <- "forestgreen"

# Gráfico
bp <- barplot(prob,
              names.arg = x,
              col = colores,
              border = "black",
              ylim = c(0, max(prob) + 0.05),
              main = "Distribución Binomial (n = 15, p = 0.4)",
              xlab = "Número de personas que se recuperan",
              ylab = "Probabilidad")

# Leyenda
legend("topright",
       legend = c(
         paste("P(X ≥ 10) =", round(p_s_10,4)),
         paste("P(3 ≤ X ≤ 8) =", round(p_3_a_8,4)),
         paste("P(X = 5) =", round(p_5,4))
       ),
       fill = c("tomato",
                "gold",
                "forestgreen"),
       bty = "n",
       cex = 0.85)
```

# Distribución de Poisson

```r
##El numero promedio de camiones-tanque que llega cada día a cierta ciudad portuaria
#es 10. Las instalaciones en el puerto pueden alojar a lo sumo 15 camiones-tanque
#por día.
#¿Cual es la probabilidad de que en un día determinado lleguen mas de 15 camiones
#y se tenga que rechazar algunos?
# Parámetro de Poisson
lambda <- 10

# Probabilidad de que lleguen más de 15 camiones
P_mas_de_15 <- 1 - ppois(15, lambda)

P_mas_de_15

# Valores posibles
x <- 0:25

# Probabilidades
prob <- dpois(x, lambda)
# Interpretación
# La probabilidad de que lleguen más de 15 camiones en un día es de 0.0486 (4,86 %).
# Esto significa que este evento es poco frecuente; sin embargo, cuando ocurre,
# el puerto supera su capacidad operativa y algunos camiones deberán esperar o ser rechazados.
# En la distribución, esta situación corresponde a las barras coloreadas en rojo,
# ubicadas a la derecha del punto de corte (15 camiones).

# Colores
colores <- ifelse(x > 15, "tomato", "steelblue")

# Gráfico
pos <- barplot(prob,
               names.arg = x,
               col = colores,
               border = "black",
               ylim = c(0, max(prob)+0.04),
               main = "Distribución de Poisson (λ = 10)",
               xlab = "Número de camiones por día",
               ylab = "Probabilidad")

# Línea de corte
abline(v = pos[which(x==15)] + diff(pos)[1]/2,
       col = "red",
       lty = 2,
       lwd = 2)

# Media de la distribución
abline(v = pos[which(x==10)],
       col = "darkgreen",
       lwd = 2)

text(pos[which(x==10)],
     max(prob)+0.02,
     expression(lambda==10),
     col="darkgreen",
     cex=0.9)

# Zona de rechazo
text(pos[which(x==20)],
     max(prob)*0.75,
     "Zona de\nrechazo",
     col="red",
     cex=0.9)

# Probabilidad sobre la zona roja
text(mean(pos[which(x>15)]),
     max(prob)*0.92,
     paste0("P(X > 15) = ", round(P_mas_de_15,4)),
     col="red",
     cex=0.9)

# Leyenda
legend("topright",
       legend=c("Operación normal (≤15)",
                "Zona de rechazo (>15)"),
       fill=c("steelblue","tomato",NA),
       border=c("black","black",NA),
       lty=c(NA,NA,1),
       lwd=c(NA,NA,2),
       col=c("black","black","darkgreen"),
       bty="n")
```

# Distribución normal

```r
###############################################
# DISTRIBUCIÓN NORMAL
###############################################

# En una fábrica de baterías, la duración (en años) sigue una
# distribución normal con media de 3 años y desviación estándar de 0.5 años.
#
# ¿Cuál es la probabilidad de que una batería dure menos de 2.3 años?

#----------------------------------------------
# Parámetros de la distribución
#----------------------------------------------

mu <- 3.0        # Media
sigma <- 0.5     # Desvío estándar
x_val <- 2.3     # Valor de interés

#----------------------------------------------
# Estandarización
#----------------------------------------------

z_val <- (x_val - mu)/sigma

cat("Valor Z =", round(z_val,2), "\n")

#----------------------------------------------
# Probabilidad utilizando el valor Z
#----------------------------------------------

pnorm(z_val)

#----------------------------------------------
# Probabilidad directamente desde la distribución normal
#----------------------------------------------

p_val <- pnorm(x_val,
               mean = mu,
               sd = sigma)

p_val

#----------------------------------------------
# Interpretación
#----------------------------------------------

cat("\nInterpretación:\n")
cat("La probabilidad de que una batería dure menos de",
    x_val,
    "años es de",
    round(p_val,4),
    "(", round(p_val*100,2),"%).\n")

cat("Esto significa que aproximadamente",
    round(p_val*100,2),
    "% de las baterías tendrán una duración inferior a",
    x_val,
    "años.\n")

###############################################
# GRÁFICO
###############################################

library(ggplot2)

# Generar datos para la curva normal

x <- seq(mu-4*sigma,
         mu+4*sigma,
         length.out=1000)

y <- dnorm(x,
           mean=mu,
           sd=sigma)

df <- data.frame(x,y)

# Crear gráfico

ggplot(df,
       aes(x,y))+

geom_line(color="blue",
          linewidth=1.2)+

# Área sombreada
geom_area(data=subset(df,x<=x_val),
          aes(x,y),
          fill="red",
          alpha=0.5)+

# Línea vertical del valor X
geom_vline(xintercept=x_val,
           colour="darkred",
           linetype="dashed",
           linewidth=1)+

# Línea vertical de la media
geom_vline(xintercept=mu,
           colour="darkgreen",
           linewidth=1)+

# Etiqueta de la media
annotate("text",
         x=mu,
         y=max(y),
         label="Media = 3",
         colour="darkgreen",
         vjust=-0.5,
         fontface="bold")+

# Etiqueta con resultados
annotate("text",
         x=2.05,
         y=max(y)*0.92,
         label=paste0(
           "X = ",x_val,
           "\nZ = ",round(z_val,2),
           "\nP(X < ",x_val,") = ",
           round(p_val,4)
         ),
         hjust=0,
         size=4)+

labs(title="Distribución Normal",
     subtitle="Área sombreada: P(X < 2.3)",
     x="Duración de las baterías (años)",
     y="Densidad")+

theme_minimal(base_size = 13)

###############################################
# INTERPRETACIÓN DEL GRÁFICO
###############################################

cat("\nInterpretación del gráfico:\n")

cat("- La curva azul representa una distribución normal con media igual a",
    mu,
    "años.\n")

cat("- La línea verde indica la media de la distribución.\n")

cat("- La línea roja punteada representa el valor X =",
    x_val,
    ".\n")

cat("- El área sombreada en rojo corresponde a la probabilidad de que una batería dure menos de",
    x_val,
    "años.\n")

cat("- Como el área sombreada representa aproximadamente",
    round(p_val*100,2),
    "% de la distribución, se concluye que es poco frecuente observar baterías cuya duración sea inferior a",
    x_val,
    "años.\n")
```

# Distribución t de Student

```r
###############################################################
# DISTRIBUCIÓN t DE STUDENT
###############################################################

# Problema:
# Se desea evaluar si un fertilizante aumenta la altura promedio
# de las plantas por encima de 18.5 cm.

###############################################################
# Datos
###############################################################

alturas <- c(19.1,18.7,20.3,18.9,19.4,20.1,
             19.8,18.6,19.7,19.0,20.2,19.3)

###############################################################
# Parámetros
###############################################################

media_muestra <- mean(alturas)
desviacion_muestra <- sd(alturas)

n <- length(alturas)

gl <- n-1

media_esperada <- 18.5

###############################################################
# Hipótesis
###############################################################

cat("Hipótesis\n")
cat("H0: μ = 18.5 cm\n")
cat("H1: μ > 18.5 cm\n\n")

###############################################################
# Estadístico t
###############################################################

t_estadistico <-
(media_muestra-media_esperada)/
(desviacion_muestra/sqrt(n))

cat("t estadístico =",round(t_estadistico,3),"\n")

###############################################################
# Valor crítico
###############################################################

alpha <- 0.05

t_critico <- qt(1-alpha,
                df=gl)

cat("t crítico =",round(t_critico,3),"\n")


###############################################################
# Decisión
###############################################################

if(t_estadistico>t_critico){

  cat("\nDecisión:\n")
  cat("Se rechaza H0.\n")
  cat("Existe evidencia estadística para afirmar que\n")
  cat("el fertilizante incrementa la altura promedio.\n")

}else{

  cat("\nDecisión:\n")
  cat("No se rechaza H0.\n")
  cat("No existe evidencia suficiente para afirmar\n")
  cat("que el fertilizante mejora el crecimiento.\n")

}

###############################################################
# GRÁFICO
###############################################################

x <- seq(-4,4,length=1000)

y <- dt(x,df=gl)

plot(x,y,
     type="l",
     lwd=2,
     col="black",
     xlab="Valores de t",
     ylab="Densidad",
     main="Distribución t de Student")

###############################################################
# Región crítica
###############################################################

xr <- seq(t_critico,4,length=200)

yr <- dt(xr,df=gl)

polygon(c(t_critico,xr,4),
        c(0,yr,0),
        col="tomato",
        border=NA)

###############################################################
# Líneas

abline(v=t_critico,
       col="red",
       lty=2,
       lwd=2)

abline(v=t_estadistico,
       col="blue",
       lwd=2)

###############################################################
# Etiquetas

text(t_critico,
     max(y)*0.85,
     "t crítico",
     col="red",
     pos=4)

text(t_estadistico,
     max(y)*0.65,
     "t observado",
     col="blue",
     pos=4)

###############################################################
# Leyenda

legend("topright",
       legend=c(
         paste("t crítico =",round(t_critico,3)),
         paste("t observado =",round(t_estadistico,3))
       ),
       col=c("red","blue","black"),
       lty=c(2,1,NA),
       lwd=c(2,2,NA),
       bty="n")

###############################################################
# Interpretación
###############################################################

cat("\nInterpretación del gráfico:\n")

cat("- La curva representa una distribución t de Student con",
    gl,
    "grados de libertad.\n")

cat("- La región roja corresponde a la región crítica de rechazo para α =",
    alpha,
    ".\n")

cat("- La línea azul representa el estadístico t calculado a partir de la muestra.\n")

cat("- Como el estadístico t se ubica dentro de la región crítica,\n")
cat("se rechaza la hipótesis nula y se concluye que el fertilizante\n")
cat("produce un aumento significativo en la altura promedio.\n")
```

# Distribución Chi cuadrado

```r
###############################################################
# DISTRIBUCIÓN CHI-CUADRADO
###############################################################

# Problema:
# Se desea determinar si existe asociación entre el sexo
# y la preferencia por un color.

###############################################################
# Datos
###############################################################

tabla <- matrix(c(20,15,15,
                  10,25,15),
                nrow=2,
                byrow=TRUE)

colnames(tabla) <- c("Rojo","Azul","Verde")
rownames(tabla) <- c("Masculino","Femenino")

tabla <- as.table(tabla)

tabla

###############################################################
# Prueba Chi-cuadrado
###############################################################

chi_resultado <- chisq.test(tabla)

print(chi_resultado)

###############################################################
# Estadístico
###############################################################

chi_valor <- chi_resultado$statistic

cat("χ² observado =",round(chi_valor,4),"\n")

###############################################################
# Grados de libertad
###############################################################

gl <- chi_resultado$parameter

cat("Grados de libertad =",gl,"\n")

###############################################################
# Valor crítico
###############################################################

alpha <- 0.05

chi_critico <- qchisq(1-alpha,
                      df=gl)

cat("χ² crítico =",round(chi_critico,4),"\n")

###############################################################
# Valor-p
###############################################################

valor_p <- chi_resultado$p.value

cat("Valor-p =",round(valor_p,5),"\n")

###############################################################
# Decisión
###############################################################

if(chi_valor>chi_critico){

cat("\nDecisión:\n")
cat("Se rechaza H0.\n")
cat("Existe evidencia estadística de asociación\n")
cat("entre el sexo y la preferencia por el color.\n")

}else{

cat("\nDecisión:\n")
cat("No se rechaza H0.\n")
cat("No existe evidencia suficiente para afirmar\n")
cat("que las variables estén asociadas.\n")

}

###############################################################
# GRÁFICO
###############################################################

x <- seq(0,20,length=1000)

y <- dchisq(x,
            df=gl)

plot(x,
     y,
     type="l",
     lwd=2,
     col="black",
     xlab=expression(chi^2),
     ylab="Densidad",
     main=expression(paste("Distribución ",chi^2)))

###############################################################
# Región crítica
###############################################################

xr <- seq(chi_critico,
          max(x),
          length=200)

yr <- dchisq(xr,
             df=gl)

polygon(c(chi_critico,xr,max(x)),
        c(0,yr,0),
        col="tomato",
        border=NA)

###############################################################
# Líneas

abline(v=chi_critico,
       col="red",
       lwd=2,
       lty=2)

abline(v=chi_valor,
       col="blue",
       lwd=2)

###############################################################
# Etiquetas

text(chi_critico,
     max(y)*0.85,
     expression(chi[critico]^2),
     pos=4,
     col="red")

text(chi_valor,
     max(y)*0.65,
     expression(chi[obs]^2),
     pos=4,
     col="blue")

###############################################################
# Leyenda

legend("topright",
       legend=c(
         paste("χ² crítico =",round(chi_critico,3)),
         paste("χ² observado =",round(chi_valor,3)),
         paste("Valor-p =",round(valor_p,5))
       ),
       col=c("red","blue","black"),
       lty=c(2,1,NA),
       lwd=c(2,2,NA),
       bty="n")

###############################################################
# Interpretación
###############################################################

cat("\nInterpretación del gráfico:\n")

cat("- La curva representa una distribución Chi-cuadrado con",
    gl,
    "grados de libertad.\n")

cat("- La región roja corresponde a la región crítica de rechazo para α =",
    alpha,
    ".\n")

cat("- La línea azul representa el estadístico χ² obtenido con los datos.\n")

if(chi_valor>chi_critico){

cat("- Como χ² observado se encuentra dentro de la región crítica,\n")
cat("se rechaza la hipótesis nula y se concluye que existe\n")
cat("una asociación significativa entre las variables.\n")

}else{

cat("- Como χ² observado no alcanza la región crítica,\n")
cat("no se rechaza la hipótesis nula y no puede afirmarse\n")
cat("que exista asociación entre las variables.\n")

}
```
