# **Construcción de la Tabla de Frecuencias Agrupadas**
Cálculo de intervalos de clase usando la regla de Sturges.

Cargamos el archivo a utilizar (airquality)

```r
data("airquality")
calidad_aire<- (airquality)
```

```r
# Calcular rango
range(calidad_aire$Wind) # 1.7 - 20.7
```

```r
# Calcular número de clases con el criterio de Sturges
k <- nclass.Sturges(calidad_aire$Wind)
k
```

```r
# Crear los intervalos de clase
Intervalo <- cut(calidad_aire$Wind, breaks = k)
```

```r
# Calcular tabla de Frecuencia Absoluta
frec_Abs <- table(Intervalo)
frec_Abs
```

```r
# Número total de datos
N <- length(calidad_aire$Wind)
N
```

```r
# Tabla de Frecuencia Relativa
frec_rel <- table(Intervalo) / N
frec_rel
```

```r
# Tabla de Frecuencia Relativa Porcentual
frec_rel_porcentual <- frec_rel * 100
frec_rel_porcentual
```

```r
# Unir todo en un Data Frame estructurado
tabla_cal <- data.frame(frec_Abs, frec_rel, frec_rel_porcentual)
tabla_cal
```

Visualizaciones y Tabla de Frecuencias con agricolae
Generación de gráficos (se mostrarán directamente debajo de la celda de Colab) e instalación del paquete agricolae.

```r
# Graficar un histograma personalizado
hist(calidad_aire$Wind,
     col = "lightgreen",
     main = "Histograma de velocidad del viento (mph)",
     xlab = "Clases",
     ylab = "Frecuencia absoluta",
     right = TRUE,
     breaks = seq(min(calidad_aire$Wind), max(calidad_aire$Wind),
                  length.out = 10),
     xlim = c(0, max(calidad_aire$Wind) + 5))
```

```r
# Instalación de agricolae para tabla de frecuencias avanzada; recuerde que en R si
# ya intalo el paquete a utilizar solo debe de activarlo usando la funcion library()
install.packages("agricolae")
library(agricolae)
```

```r
# Gráfico de frecuencias e impresión de la tabla agrupada
frec_calidad_aire <- with(calidad_aire,
                          graph.freq(calidad_aire$Wind, plot = TRUE))
```

```r
tabla_calidad_viento <- print(table.freq(frec_calidad_aire),
                              row.names = FALSE)
```
