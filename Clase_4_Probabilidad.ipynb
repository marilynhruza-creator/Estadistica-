## Probabilidad clásica

```r
# Sucesos opuestos
prob_A <- 1/6
prob_A_opuesto <- 1-prob_A
cat("Probabilidad de no obtener un 4:", prob_A_opuesto,"\n")
```

```r
# Sucesos mutuamente excluyentes
prob_impar <- 3/6
prob_par <- 3/6
prob_i_p <- prob_impar + prob_par
cat("Probabilidad impar o par:", prob_i_p,"\n")
```

```r
# Sucesos no mutuamente excluyentes
total_estudiantes <- 30
estudiantes_con_telefono <- 18
estudiantes_con_reloj <- 10
estudiantes_con_telefono_y_reloj <- 2

prob_t <- estudiantes_con_telefono/total_estudiantes
prob_r <- estudiantes_con_reloj/total_estudiantes
prob_t_y_r <- estudiantes_con_telefono_y_reloj/total_estudiantes
prob_t_r <- prob_t + prob_r - prob_t_y_r
cat("Teléfono o reloj:", prob_t_r,"\n")
```

```r
# Independientes
total_clientes <- 500
hombres <-130
mujeres <-370
prob_H <- hombres/total_clientes
prob_M_H <- mujeres/(total_clientes-1)
prob_H_y_M <- prob_H*prob_M_H
cat("Hombre y luego mujer:", prob_H_y_M,"\n")

prob_H_H <- prob_H*((hombres-1)/(total_clientes-1))
cat("Dos hombres:", prob_H_H,"\n")

```

```r
# Condicional
total_estudiantes <-100
estudiantes_que_estudiaron <-80
estudiantes_que_pasaron_y_estudiaron <-60
prob_A_B <- estudiantes_que_pasaron_y_estudiaron/total_estudiantes
prob_B <- estudiantes_que_estudiaron/total_estudiantes
prob_condicional <- prob_A_B/prob_B
cat("P(A|B):", prob_condicional,"\n")
```

## Probabilidad frecuencial

```r
n<-9
frec_B<-4; frec_G<-3; frec_N<-2
cat(frec_B/n, frec_G/n, frec_N/n)
```

## Carga de datos externos a R

```r
url <- "https://github.com/hllinas/DatosPublicos/raw/main/Estudiantes.Rdata"
download.file(url,"Estudiantes.Rdata",mode="wb")
load("Estudiantes.Rdata")

variables <- c("Edad","P1","P2","P3","Final","Definitiva","Gastos","Ingreso","Gas","Puntaje")
Estudiantes[variables] <- lapply(Estudiantes[variables], as.numeric)

str(Estudiantes)
head(Estudiantes)
```

## Ejercicios

```r
n_total <- nrow(Estudiantes)
n_femenino <- sum(Estudiantes$Sexo=="Femenino")
cat("P(Femenino):", n_femenino/n_total,"\n")

n_privado <- sum(Estudiantes$Colegio=="Privado")
n_alta <- sum(Estudiantes$Estatura=="Alta")
n_inter <- sum(Estudiantes$Colegio=="Privado" & Estudiantes$Estatura=="Alta")
cat("P(Privado o Alta):",(n_privado+n_alta-n_inter)/n_total,"\n")

priv <- Estudiantes[Estudiantes$Colegio=="Privado",]
cat("P(Definitiva>4|Privado):",sum(priv$Definitiva>4)/nrow(priv),"\n")
```
